## Request LLM through LangGraph

```python
from langchain.agents import create_agent
from config.load_key import load_key
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model=load_key("openai", "model"),
    api_key=load_key("openai", "api_key"),
)

agent = create_agent(
    model=model,
    tools=[],
    system_prompt="You are a helpful assistant.",
)

agent.invoke({"messages": [{"role": "user", "content": "Who are you? What problem can you help me with?"}]})
```
