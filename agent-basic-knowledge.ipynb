## Provide tools for LLM through LangChain

```python
import datetime
from langchain.tools import tool
from config.load_key import load_key
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model=load_key("openai", "model"),
    api_key=load_key("openai", "api_key"),
)


@tool
def get_current_date():
    """Return the current date."""
    return datetime.datetime.today().strftime("%Y-%m-%d")


llm_with_tools = llm.bind_tools([get_current_date])
all_tools = {"get_current_date": get_current_date}

query = "When is it today?"
messages = [query]

ai_msg = llm_with_tools.invoke(messages)
print(ai_msg)
messages.append(ai_msg)

print(ai_msg.tool_calls)
if ai_msg.tool_calls:
    for tool_call in ai_msg.tool_calls:
        selected_tool = all_tools[tool_call["name"].lower()]
        tool_msg = selected_tool.invoke(tool_call)
        messages.append(tool_msg)

llm_with_tools.invoke(messages).content
```

## Provide tools for LLM through LangGraph

```python
from langchain.agents import create_agent
from config.load_key import load_key
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model=load_key("openai", "model"),
    api_key=load_key("openai", "api_key"),
)


@tool
def get_current_date():
    """Return the current date."""
    return datetime.datetime.today().strftime("%Y-%m-%d")


agent = create_agent(
    model=llm,
    tools=[get_current_date],
    system_prompt="You are a helpful assistant.",
)

agent.invoke({"messages": [{"role": "user", "content": "When is it today?"}]})
```

```python
from typing import TypedDict, Annotated
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, END
from langgraph.graph.message import add_messages
from langgraph.prebuilt import ToolNode, tools_condition
from config.load_key import load_key

# 1. Initialize the LLM
llm = ChatOpenAI(
    model=load_key("openai", "model"),
    api_key=load_key("openai", "api_key"),
)


# 2. Define a tool
@tool("devide")
def devide(a: int, b: int) -> float:
    """
    Calculate the quotient of a and b.
    Args:
        a (int): The numerator.
        b (int): The denominator.
    """
    if b == 0:
        raise ZeroDivisionError("Division by zero is not allowed.")
    elif b == 1:
        raise ValueError("Cannot divide by 1.")
    return a / b


# 3. Tool error handler
def handle_tool_error(error: Exception) -> str:
    if isinstance(error, ValueError):
        return "Cannot divide by 1."
    elif isinstance(error, ZeroDivisionError):
        return "Division by zero is not allowed."
    return f"Tool error: {error}"


# 4. Bind tools to the model so it can emit tool calls
llm_with_tools = llm.bind_tools([devide])


# 5. Define the graph state (only track messages)
class State(TypedDict):
    messages: Annotated[list, add_messages]


# 6. Agent node: calls the LLM
def agent_node(state: State):
    msg = llm_with_tools.invoke(state["messages"])
    return {"messages": [msg]}


# 7. Tool node: executes tool calls and handles errors
tool_node = ToolNode(
    tools=[devide],
    handle_tool_errors=handle_tool_error,
)

# 8. Build the graph
graph = StateGraph(State)
graph.add_node("agent", agent_node)
graph.add_node("tools", tool_node)
graph.set_entry_point("agent")

# If the agent produced tool calls → go to tools, otherwise end
graph.add_conditional_edges("agent", tools_condition, {"tools": "tools", END: END})

# After tool execution, return control to the agent
graph.add_edge("tools", "agent")

# 9. Compile the graph
app = graph.compile()

result = app.invoke(
    {"messages": [HumanMessage(content="What is 10 divided by 1?")]}
)

print(result["messages"][-1].content)
```
